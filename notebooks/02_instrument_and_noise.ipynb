# Notebook 02: Simulated Instrument & Noise Modeling
Simulates laser ramp tuning, optical etalon interference, colored 1/f noise, and ADC quantization.

```
import numpy as np
import matplotlib.pyplot as plt
from tdlas.hitran_io import load_hitran_ch4_window
from tdlas.instrument import DeviceParams, simulate_scan

lines = load_hitran_ch4_window(3017.0, 3019.0)
dev = DeviceParams.sample_random("DEV_NOTEBOOK", seed=42)
rng = np.random.default_rng(42)
scan = simulate_scan(x_ppm=25.0, T_K=296.0, P_atm=1.0, device=dev, rng=rng, lines_data=lines, include_noise=True)

fig, ax = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
ax[0].plot(scan["time_s"] * 1000, scan["voltage_v"], "r-", label="Detector Voltage V(t)")
ax[0].plot(scan["time_s"] * 1000, scan["ground_truth"]["laser_baseline_v"], "k--", label="Laser Power Baseline")
ax[0].set_ylabel("Voltage [V]")
ax[0].set_title("Simulated Raw Scan with Etalon, 1/f Noise, and 16-bit ADC")
ax[0].legend()
ax[0].grid(True)

ax[1].plot(scan["time_s"] * 1000, scan["etalon_channel_v"], "b-", label="Reference Etalon Channel")
ax[1].set_xlabel("Time [ms]")
ax[1].set_ylabel("Etalon Voltage [V]")
ax[1].legend()
ax[1].grid(True)
plt.tight_layout()
plt.show()
```

### Noise Breakdown
- **Gaussian White Noise:** Detector thermal and shot noise.
- **1/f Flicker Noise:** Generated via FFT spectral shaping with 1/f slope.
- **Baseline Drift:** Thermal drift modeled as a slow random walk.
