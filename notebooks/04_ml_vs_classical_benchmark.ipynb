# Notebook 04: Machine Learning vs Classical Inversion Under Distribution Shifts
Classical Levenberg-Marquardt vs a 1D CNN and a 5-member deep ensemble, trained on device-split simulated data and evaluated on six shifted scenarios using held-out test devices. The numbers come from `scripts/run_ml_benchmark.py`; this notebook only reads the stored result.

```
import json
from pathlib import Path

result = json.loads(Path('../docs/results/ml_benchmark.json').read_text())
cfg = result['config']
print(f"{len(cfg['train_devices'])} train / {len(cfg['val_devices'])} val / {len(cfg['test_devices'])} test devices")
for name, r in result['scenarios'].items():
    print(f'=== {name} ===')
    for key, label in (('classical', 'LM fit'), ('ml_single', 'CNN'), ('ml_ensemble', 'Ensemble')):
        m = r[key]
        print(f"  {label:<9} RMSE {m['rmse_ppm']:.2f} ppm | bias {m['bias_ppm']:+.2f} | 95% coverage {m['coverage_95']:.1%} | NLL {m['nll']:.2f}")
```

### Reading the results
Read the stored table, not a claim: where the classical fit or the ensemble wins, and how well each 95% interval is calibrated, is shown by the output above and by `docs/figures/fig2_physics_vs_ml_comparison.png`. The CNN is small and was tuned only over three learning rates on the validation devices, so a negative result for ML here does not bound what a better-tuned network could do.
