# Notebook 01: Physical Forward Model & Voigt Profile Validation
Demonstrates line parameters, partition functions, and the from-scratch Weideman rational Voigt lineshape compared with SciPy.

```
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import wofz
from tdlas.hitran_io import load_hitran_ch4_window, partition_function_ch4
from tdlas.lineshape import faddeeva_rational, voigt_profile
from tdlas.absorption import compute_absorption_coefficient, compute_transmission

print("Modules loaded successfully.")
```

## 1. HITRAN CH4 Line Positions and TIPS Partition Function

```
lines = load_hitran_ch4_window(3017.0, 3019.0)
temps = np.linspace(200.0, 400.0, 50)
Q_vals = [partition_function_ch4(T) for T in temps]

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(temps, Q_vals, "b-", lw=2)
ax[0].set_title("Total Internal Partition Sum Q(T)")
ax[0].set_xlabel("Temperature [K]")
ax[0].set_ylabel("Q(T)")
ax[0].grid(True)

nu_centers = [l["nu_cm1"] for l in lines]
intensities = [l["S_T0"] for l in lines]
ax[1].stem(nu_centers, intensities)
ax[1].set_title("HITRAN CH4 Line Strengths at 296 K")
ax[1].set_xlabel("Wavenumber [cm⁻¹]")
ax[1].set_ylabel("S(T0) [cm⁻¹/(molec·cm⁻²)]")
ax[1].grid(True)
plt.tight_layout()
plt.show()
```

## 2. From-Scratch Voigt Profile vs SciPy Benchmark

```
nu_axis = np.linspace(3017.0, 3019.0, 2000)
absorbance, transmission = compute_transmission(nu_axis, x_ppm=50.0, T_K=296.0, P_atm=1.0, L_cm=70.0, lines_data=lines)

plt.figure(figsize=(8, 4))
plt.plot(nu_axis, transmission, "g-", lw=2, label="Transmission T(ν) (50 ppm, 70 cm)")
plt.title("Beer-Lambert Optical Transmission of CH4 at 3313 nm")
plt.xlabel("Wavenumber [cm⁻¹]")
plt.ylabel("I(ν) / I₀(ν)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
